```python
##5. What is the average length of stay (admission to discharge)?
```

```python
df['Date_of_Admission'] = pd.to_datetime(df['Date_of_Admission'])
```

```python
df['Date_of_Discharge'] = pd.to_datetime(df['Date_of_Discharge'])
```

```python
df['Length of Stay(Days)'] = (df['Date_of_Discharge'] - df['Date_of_Admission']).dt.days.astype('Int64')
```
